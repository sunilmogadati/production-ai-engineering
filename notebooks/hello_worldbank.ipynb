## 2. Why log the income? (compare the fit both ways)

```python
y = df["life_expectancy"]
r2_raw = LinearRegression().fit(df[["gdp_per_capita"]], y).score(df[["gdp_per_capita"]], y)
r2_log = LinearRegression().fit(np.log(df[["gdp_per_capita"]]), y).score(np.log(df[["gdp_per_capita"]]), y)
print(f"fit on RAW income : R^2 = {r2_raw:.3f}")
print(f"fit on LOG income : R^2 = {r2_log:.3f}   <- the log nearly doubles it")
```

## 3. Fit the model + the two scores

```python
def adjusted_r2(r2, n, p): return 1 - (1-r2)*(n-1)/(n-p-1)
X = np.log(df[["gdp_per_capita"]]); n = len(df)
model = LinearRegression().fit(X, y)
t0, t1, r2 = model.intercept_, model.coef_[0], model.score(X, y)
print(f"life_expectancy = {t0:.1f} + {t1:.1f} * ln(GDP per capita)")
print(f"R-squared          = {r2:.3f}   income explains {r2*100:.0f}% of the variation")
print(f"Adjusted R-squared = {adjusted_r2(r2, n, 1):.3f}")
```

## 4. Plot it

```python
plt.figure(figsize=(8,5))
plt.scatter(df.gdp_per_capita, df.life_expectancy, s=25, alpha=.6)
xs = np.linspace(df.gdp_per_capita.min(), df.gdp_per_capita.max(), 200)
plt.plot(xs, t0 + t1*np.log(xs), "r", lw=2, label=f"life = {t0:.1f} + {t1:.1f}·ln(GDP)")
plt.xscale("log"); plt.xlabel("GDP per capita (log scale)"); plt.ylabel("life expectancy")
plt.legend(); plt.title("Richer countries live longer"); plt.show()
```

## 5. Diminishing returns — every *doubling* of income buys the same ~3 years

```python
for lo, hi in [(1_000, 2_000), (4_000, 8_000), (50_000, 100_000)]:
    plo, phi = t0 + t1*np.log(lo), t0 + t1*np.log(hi)
    print(f"${lo:>7,} -> ${hi:>7,}/person : {plo:.1f} -> {phi:.1f} yrs  (+{phi-plo:.1f})")
```

## 6. The residuals — who beats their income, who falls short (the real story)

```python
df["predicted"] = model.predict(X)
df["surprise"] = df.life_expectancy - df.predicted
print("LIVE LONGER than income predicts:")
print(df.nlargest(4,"surprise")[["country","life_expectancy","predicted","surprise"]].to_string(index=False))
print("\nLIVE SHORTER than income predicts:")
print(df.nsmallest(4,"surprise")[["country","life_expectancy","predicted","surprise"]].to_string(index=False))
```

## Takeaway
- Same linear regression, now on 210 real countries.
- The **log** straightened a curved relationship.
- One feature, **R²=0.71** — income explains most, not all.
- **Correlation, not cause** — income buys the things that add years.
- The **residuals** are often the real story.
